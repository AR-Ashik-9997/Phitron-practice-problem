```python
import torch
```

```python
X=torch.tensor([2.0],requires_grad=True)
y=X**2+3*X+1
y.backward()
```

PyTorch breaks down the expression $y = x^2 + 3x + 1$ into primitive elementary ops:
- $u = x^2 \implies 2.0^2 = 4.0$ (Operation: PowBackward0)
- $v = 3 \cdot x \implies 3 \times 2.0 = 6.0$ (Operation: MulBackward0)
- $w = u + v \implies 4.0 + 6.0 = 10.0$ (Operation: AddBackward0)
- $y = w + 1 \implies 10.0 + 1 = 11.0$ (Operation: AddBackward0)
