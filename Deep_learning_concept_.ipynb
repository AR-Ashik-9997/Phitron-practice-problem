```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
X = torch.tensor([[1.0, 2.0], [2.0, 3.0]], dtype=torch.float32)
y_true = torch.tensor([[1.0], [0.0]], dtype=torch.float32)

print("--- Step 1: Input Data ---")
print("Input Features (X):\n", X)
print("True Targets (y):\n", y_true)
```

```python
class SimpleNeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        # Input: 2, Hidden: 3, Output: 1
        self.hidden_layer = nn.Linear(in_features=2, out_features=3)
        self.activation = nn.ReLU()  # Activation Function
        self.output_layer = nn.Linear(in_features=3, out_features=1)
        self.sigmoid = nn.Sigmoid()  # Output Activation (Probability)

    def forward(self, x):
        h = self.activation(self.hidden_layer(x))
        out = self.sigmoid(self.output_layer(h))
        return out

model = SimpleNeuralNetwork()

print("\n--- Step 2: Model Architecture ---")
print(model)
```

```python
# Forward Propagation and Loss Function
criterion = nn.MSELoss()
y_pred = model(X)
loss = criterion(y_pred, y_true)

print("\n--- Step 3: Forward Pass & Initial Loss ---")
print("Initial Predictions (y_pred):\n", y_pred.detach())
print("Initial Loss Value:", loss.item())
```

```python
# Backpropagate gradients
loss.backward()
print("\n--- Step 4: Backpropagation (Gradients) ---")
print("Hidden Layer Weight Gradients:\n", model.hidden_layer.weight.grad)
```

```python
torch.manual_seed(42)

# Optimizer
optimizers_dict = {
    "1. Standard SGD": lambda params: optim.SGD(params, lr=0.1),
    "2. SGD with Momentum": lambda params: optim.SGD(params, lr=0.1, momentum=0.9),
    "3. RMSprop": lambda params: optim.RMSprop(params, lr=0.1),
    "4. Adam": lambda params: optim.Adam(params, lr=0.1)
}

criterion = nn.MSELoss()

print("--- Comparison of Different Optimizers ---\n")

for name, opt_func in optimizers_dict.items():
    torch.manual_seed(42)
    model = SimpleNeuralNetwork()
    optimizer = opt_func(model.parameters())

    # Forward Pass & Loss
    y_pred = model(X)
    loss = criterion(y_pred, y_true)

    # Backpropagation
    optimizer.zero_grad()
    loss.backward()

    # befor older weight
    initial_weights = model.hidden_layer.weight.clone().detach()

    # Optimizer update
    optimizer.step()

    # updated after weight
    updated_weights = model.hidden_layer.weight.clone().detach()

    print(f"=== {name} ===")
    print("Initial Loss:", round(loss.item(), 5))
    print("\n\n Weights BEFORE update:", initial_weights.numpy())
    print("\n Weights AFTER update:", updated_weights.numpy())
    print("-" * 45)
```
