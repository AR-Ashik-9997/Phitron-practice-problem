```python
X = torch.tensor([[1.0, 2.0]])
W1 = torch.tensor([[0.5, -0.2, 0.1], [0.3, 0.8, -0.5]])
b1 = torch.tensor([[0.1, 0.0, -0.1]])
W2 = torch.tensor([[0.7], [0.2], [-0.4]])
b2 = torch.tensor([[0.05]])

Z1=torch.matmul(X,W1)+b1
A1=torch.relu(Z1)
Z2=torch.matmul(A1,W2)+b2

print("Layer 1 Pre-activation (Z1):", Z1)
print("Layer 1 Post-activation (A1):", A1)
print("Final Output (Z2):", Z2)
```

```python
X=torch.tensor([[1.0],[2.0]])
W1=torch.tensor([[0.1,0.2],[0.3,0.4]])
b1 = torch.tensor([[0.1,-0.1]])
W2=torch.tensor([[0.7,0.4]])
b2=torch.tensor([[0.05]])
```

```python
Z1=torch.matmul(W1,X)+b1
A1=torch.relu(Z1)
y_hat=torch.matmul(W2,A1)+b2

print(f"Z1:\n{Z1}")
print(f"\nA1:\n{A1}")
print(f"\ny_hat:\n{y_hat}")
```
